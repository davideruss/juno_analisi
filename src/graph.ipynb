```python
from pathlib import Path
import uproot
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc
```

```python
root_path = Path("fit/fit_run_14383.root")

with uproot.open(root_path) as file:
    fits = file["fit_summary"].arrays(library="np")
    histograms = {}
    for bin_id in fits["bin_id"]:
        h = file[f"h_{int(bin_id)}"]
        counts, edges = h.to_numpy(flow=False)
        histograms[int(bin_id)] = (counts, edges, h.errors(flow=False))

print(f"Run {fits['run'][0]}: {len(histograms)} istogrammi")
```

```python
def exp_gauss(t, area, mu, sigma, tau):
    """Riproduce anche il ramo asintotico usato in fit_runs.cpp."""
    t = np.asarray(t, dtype=float)
    if sigma <= 0 or tau <= 0:
        return np.full_like(t, np.nan)
    a = (mu + sigma*sigma/tau - t) / (np.sqrt(2.)*sigma)
    exponent = (mu-t)/tau + sigma*sigma/(2.*tau*tau)
    y = np.empty_like(t)
    mask = a > 20.
    u = 1./(2.*a[mask]**2)
    scaled_erfc = (1.-u+3.*u*u-15.*u**3+105.*u**4)/(np.sqrt(np.pi)*a[mask])
    y[mask] = area/(2.*tau)*np.exp(-0.5*((t[mask]-mu)/sigma)**2)*scaled_erfc
    y[~mask] = area/(2.*tau)*np.exp(exponent[~mask])*erfc(a[~mask])
    return y


def double_exp_gauss(t, params):
    C1, mu, sigma, tau1, C2, tau2, background = params
    return (exp_gauss(t, C1, mu, sigma, tau1)
            + exp_gauss(t, C2, mu, sigma, tau2) + background)
```

```python
# fit_runs.cpp usa 10 bin angolari: la griglia si adatta al numero presente.
ncols = 4
nrows = 5
fig, axes = plt.subplots(nrows, ncols, figsize=(20, 4*nrows),
                         squeeze=False, constrained_layout=True)
fig.suptitle(f"Run {fits['run'][0]} — DoubleExpGauss", fontsize=17)

x_per_dopo =[]
a = []
b=[]
for row, ax in enumerate(axes.flat):

    if row >= len(fits["bin_id"]):
        ax.set_axis_off()
        continue

    bin_id = int(fits["bin_id"][row])

    y, edges, error = histograms[bin_id]

    centers = (edges[:-1] + edges[1:])/2

    ax.stairs(y, edges, color="tab:blue", label="Dati")

    ax.errorbar(centers, y, yerr=error, fmt="none",
                color="tab:blue", alpha=0.3, linewidth=0.6)
    
    title = (f"Bin {bin_id}: cos(theta) in "
             f"[{fits['cos_low'][row]:.2f}, {fits['cos_high'][row]:.2f}]")
    if fits["accepted"][row]:
        xx = np.linspace(fits["fit_min"][row], fits["fit_max"][row], 1500)
        with np.errstate(over="ignore", invalid="ignore"):
            yy = double_exp_gauss(xx, fits["params"][row])
        nonfinite = ~np.isfinite(yy)
        yy[nonfinite] = np.nan
        status = int(fits["status"][row])
        clean = status == 0 and bool(fits["root_valid"][row])
        ax.plot(xx, yy, color="tab:red", linestyle="-", # if clean else "--",
                label="DoubleExpGauss")# if clean else "Fit accettato con riserva")
        ax.axvline(fits["peak"][row], color="0.4", linestyle=":", linewidth=1)
        info = (rf"$\chi^2 = {fits['chi2'][row]:.2f}$" + "\n"
                + rf"ndf = {int(fits['ndf'][row])}; $\chi^2/ndf = {fits['chi2_ndf'][row]:.2f}$"
                + f"\nPicco = {fits['peak'][row]:.3f} ns"
                + f"\nstatus = {status}, IsValid = {bool(fits['root_valid'][row])}")

        if nonfinite.any():

            info += "\nCurva con valori non finiti"
        ax.text(0.97, 0.97, info, transform=ax.transAxes, ha="right", va="top",
                fontsize=9, bbox=dict(facecolor="white", edgecolor="0.8", alpha=0.9))
    else:
        title += "\nNessun tentativo accettato"
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("Tempo residuo [ns]")
    ax.set_ylabel("Conteggi")
    ax.set_yscale('log')
    ax.grid(alpha=0.2)
    ax.legend(loc="upper left", fontsize=8)

# Per salvare anche la figura, togli il commento:
# fig.savefig(f"grafici_run_{fits['run'][0]}.png", dpi=200)
plt.show()
```
