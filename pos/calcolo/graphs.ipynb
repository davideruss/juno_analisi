```python
import numpy as np 
import matplotlib.pyplot as plt

y, z, t = np.loadtxt("output16000_0.dat", unpack=True, usecols= (2, 3, 4))

plt.figure(figsize=(10, 6))
map = plt.scatter(y, z, c=t, cmap='turbo_r', s=10, label= "pmt")

cbar = plt.colorbar(map)
cbar.set_label('Tempo di volo [s]', fontsize=12)

plt.scatter(0, 16000, color='red', marker='D', s=100, label='Sorgente', zorder=0)

plt.xlabel('Y [mm]')
plt.ylabel('Z [mm]')
plt.title('Tempo di arrivo dei fotoni sui PMT (Sezione X ≈ 0)', fontsize=14)

plt.gca().set_aspect('equal')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

r= 17700
theta= np.linspace(0, 2*np.pi, 100)
y= r*np.sin(theta)
x=r* np.cos(theta)

a, b= np.loadtxt("pos1.csv", usecols=(2,3), unpack= True)

plt.plot(x, y, 'r-', linewidth=3, label= "confine LS-W")
plt.scatter(a,b, s=15, label= "Posizioni pmt")

plt.gca().set_aspect('equal')
plt.legend()
plt.show()
```
