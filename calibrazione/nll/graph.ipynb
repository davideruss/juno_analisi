## Librerie e coordinate


```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
z_source = np.loadtxt("output_con_picchi.txt", usecols=1)
theta=np.linspace(-0.975, 0.975, 40)
print(theta)
```

## Mappa nel quadrante radiale positivo


```python
z_zoom = z_source[18:]
theta_zoom= theta[20:]

delta = np.loadtxt("risultati/delta_r.txt", unpack= True) 

print(theta_zoom.shape)

fig, ax = plt.subplots(figsize=(15, 7))

plot = ax.pcolormesh(
    z_zoom,
    theta_zoom,
    delta,
    cmap="seismic",
    alpha = 0.8,
    vmin=-10,
    vmax=10
)

cbar = fig.colorbar(plot, ax=ax)
cbar.set_label(r"$\Delta t$ [ns]")

ax.set_xlabel("posizione radiale della sorgente r[mm]", fontsize=13)
ax.set_ylabel(r"$\cos(\theta)$", fontsize=13)
ax.set_title(r"$\Delta t$ in funzione di $\cos(\theta)$ e r", fontsize=15)
ax.grid(True, linestyle='--', alpha=0.6)   
plt.tight_layout()
plt.show()
```
