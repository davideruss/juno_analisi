# 0 · Bias iniziale
Si fitta con una gaussiana la distribuzione della posizione ricostruita:
**bias = μ − z della sorgente**, **risoluzione = σ**.
Il campione è selezionato entro 1 m dalla sorgente e nel picco energetico ±100 NPE.

La tabella contiene i risultati iniziali di `~/juno/bias/output6.root`:
39 posizioni, prima del confronto con le correzioni. Gli errori verticali sono quelli dei fit.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

dati = Path("dati")
plt.rcParams.update({"font.size": 11, "axes.spines.top": False,
                     "axes.spines.right": False, "figure.dpi": 120})

z, bias, err_bias, sigma, err_sigma = np.loadtxt(
    dati / "bias_iniziale.csv", delimiter=",", skiprows=1, unpack=True)
assert len(z) == 39 and np.all(np.diff(z) > 0)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3.7), sharex=True, layout="constrained")
ax[0].errorbar(z / 1000, bias, yerr=err_bias, fmt="o", ms=4, color="#275a8e")
ax[0].axhline(0, color="0.5", lw=1, ls="--")
ax[0].set(title="Bias iniziale", ylabel=r"$\mu-z_{sorgente}$ [mm]")
ax[1].errorbar(z / 1000, sigma, yerr=err_sigma, fmt="o", ms=4, color="#c3632b")
ax[1].set(title="Risoluzione lungo z", ylabel=r"$\sigma$ [mm]")
for a in ax:
    a.set_xlabel("Posizione della sorgente z [m]")
    a.grid(alpha=0.2)
plt.show()
```

Il bias raggiunge circa **173 mm in valore assoluto**, mentre σ varia tra **197 e 276 mm**. Questo è il riferimento iniziale per lo studio dei tempi di volo.
